After our initial experiments with unsupervised analysis of the `10th September` dataset, we are taking a step back and examining the text encoding (vectorisation) or embedding models.

### Embedding

A typical first step in unsupervised text analysis is to transform textual data into numerical representations (embeddings). These vector representations capture semantic information and can then be used to cluster texts according to their **semantic similarity**.


---

### Capturing Semantics

Text encoding has evolved significantly, from early approaches based on word counts and co-occurrence statistics to static word embeddings and, more recently, contextual embeddings produced by transformer-based models.

While earlier methods primarily captured lexical and distributional relationships between words, contextual models generate representations that account for the surrounding text, allowing the same word or token to have different representations depending on its context.

This limitation becomes particularly relevant when analysing short texts, where limited linguistic context makes semantic interpretation more challenging. For example, X (formerly Twitter) is a microblogging platform where standard posts are limited to 280 characters. The challenge is even greater when considering other contextual elements, such as users' self-descriptions or biographies. These can be extremely short, sometimes consisting of only one or two tokens, and may include non-linguistic elements such as emojis representing national flags, religious symbols, or emotions. In such cases, interpreting meaning requires going beyond the textual content alone and considering the broader context in which it occurs.
